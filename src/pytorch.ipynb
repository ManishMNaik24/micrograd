```python
import torch
```

```python
x1=torch.Tensor([2.0]).double()  ; x1.reqires_grad=True
x2=torch.Tensor([0.0]).double()  ; x2.reqires_grad=True
w1=torch.Tensor([-3.0]).double()  ; w1.reqires_grad=True
w2=torch.Tensor([1.0]).double()  ; w2.reqires_grad=True

b=torch.Tensor([6.8813735870195432]) ; b.requires_grad=True
n=x1*w1+x2*w2+b
o=torch.tanh(n)

print(o.data)
```
